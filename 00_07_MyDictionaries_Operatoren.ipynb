# Operatoren Dictionaries
- `len(d)` - Länge ermitteln
- `d[k]` - Zugreifen auf Schlüssel (k)
- `del d[k]` - Löschen des Schlüssels und seines Wertes
- `k in d` - Überprüfen, ob Schlüssel vorhanden
- `k not in d` - Überprüfen, ob Schlüssel NICHT vorhanden
- `d.get('...') is None` - Überprüfen, ob Wert NICHT vorhanden
- `d1 | d2` - Zwei Dictionaries kombinieren


```python
woerterbuch = {
    'Germany': 'Deutschland',
    'France': 'Frankreich',
    'Italy': 'Italien',
    'Spain': 'Spanien',
    'Portugal': 'Portugal'
}
len(woerterbuch)
```

```python
woerterbuch['Italy']
```

```python
del woerterbuch['Portugal']
```

```python
# ACHTUNG: Der nachfolgende Befehl löscht das GESAMTE Wörterbuch!
del woerterbuch
```

```python
woerterbuch['Portugal']
```

```python
if woerterbuch.get('Portugal') is None:
    print("Das Wort 'Portugal' ist NICHT im Wörterbuch enthalten.")
```

```python
if woerterbuch.get('Portugal'):
    print("Das Wort 'Portugal' ist im Wörterbuch enthalten.")
else:
    print("Das Wort 'Portugal' ist NICHT im Wörterbuch enthalten.")
```

```python
# Auf bestimmten Schlüssel prüfen
if 'Portugal' in woerterbuch:
    print("Das Wort 'Portugal' ist im Wörterbuch enthalten.")
else:
    print("Das Wort 'Portugal' ist NICHT im Wörterbuch enthalten.")
```

```python
# Zwei Wörterbücher zusammenführen - erst ab Python 3.9 möglich
woerterbuch = {
    'Germany': 'Deutschland',
    'France': 'Frankreich',
    'Italy': 'Italien',
    'Spain': 'Spanien',
    'Portugal': 'Portugal'
}
print("Wörterbuch 1:", len(woerterbuch))
woerterbuch2 = {
    "Columbia": "Kolumbien",
    "Russia": "Russland",
    "Polen": "Polen"
}
print("Wörterbuch 2:", len(woerterbuch2))
new_woerterbuch = woerterbuch | woerterbuch2
new_woerterbuch
print("Neues Wörterbuch:", len(new_woerterbuch))
```
