## Arten von Zahlen

Python beinhaltet verschiedene "Arten" von Zahlen. 
Integers sind einfach ganze Zahlen, positiv und negativ. Zum Beispiel: 2 und -2 sind Integers.
Floats sind Kommazahlen und in Python besonders, da sie einen Dezimalpunkt oder einen Exponenten (e) beinhalten, um die Zahl zu definieren. Zum Beispiel sind 2.0 und -2.1 Zahlen mit Dezimalpunkt. 4E2 (4 mal 10 hoch 2) ist ebenso ein Beispiel für eine Float-Zahl in Python.

## Grundlegende Arithmetik

```python
# Addition
2 + 2
```

```python
# Subtraktion
4 - 2
```

```python
# Multiplikation
2 * 3
```

```python
# Division 
4 / 2  # Ergebnis ist ein Float-Wert
```

```python
# Integerdivision
7 // 4  # Dezimalstellen werden ohne Rundung abgeschnitten
```

```python
# Modulo - Den Rest der Division ermitteln
7 % 4
```

```python
# Potenz
2 ** 3
```

```python
# Wurzel ziehen
4 ** 0.5
```

```python
# Reihenfolge der Operatoren wird in Python eingehalten
2 + 10 * 10 + 3
```

```python
# Klammern 
(2 + 10) * (10 + 3)
```
