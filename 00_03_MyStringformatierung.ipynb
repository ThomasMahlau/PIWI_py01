<h1 style='color:#4876FF'><b>Stringformatierungen</b></h1>

#### Varianten:
- Verbindung mit `+` (Konkatenieren)
- mit `Modulo`-Zeichen
- mit `Format`-Methode
- mit `f`-Strings (String-Literale)

### Konkatenieren

```python
# Einfache Kombination funktioniert nicht, da int nicht zu str passt.
spieler = 'Max'
punkte = 120
'Heute hat ' + spieler + 'eine Punktzahl von ' + punkte + 'erreicht'
```

```python
# Datentypumwandlung mit str()
spieler = 'Max'
punkte = 120
'Heute hat ' + spieler + ' eine Punktzahl von ' + str(punkte) + ' erreicht'
```

<h2 style='color: 	#b78900;'>Modulo-Zeichen</h2>

```python
# %s als 'Stringformatierungsoperator'
print("Hier werde ich %s einfügen." % 'etwas')
```

```python
# mehrere Elemente (Tupel) einfügen
print("Ich werde hier %s Text einfügen und einen %s hier." % ('EINEN','WEITEREN'))
```

```python
# Variablennamen übergeben
a = "einen"
b = "weiteren"
print("Ich werde hier %s Text einfügen und einen %s hier." % (a, b))
```

```python
a, b = "einen", "weiteren" # [mehrere Variablen in einer Anweisung definieren]
print("Ich werde hier %s Text einfügen und einen %s hier." % (a, b))
```

```python
# %s nutzt Methode str()
# %r nutzt Methode repr() - Stringrückgabe INKLUSIVE Hochkomma und Markierungsanweisungen ("representation" eines Objektes)
print('Er sagte, sein Name sei %s.' % 'Fred')
print('Er sagte, sein Name sei %r.' % 'Fred')
```

```python
# %s konvertiert alles in String - inklusive int und float
# %d konvertiert Zahlen in int - ohne Rundung
print('Ich schrieb heute %s Programme.' % 3.75)
print('Ich schrieb heute %d Programme.' % 3.75)
```

```python
# f-Strings (ab Python 3.6)
name = 'Max'
age = 28
print(f'{name} ist {age} Jahre alt.')
```

<h2 style='color: 	#b78900;'>Einrückung und Präzision von Gleitkommazahlen</h2>

```python
# Format %5.2f
# 5 = minimale Anzahl Zeichen - wenn weniger, wird durch Leerzeichen aufgefüllt
# .2f = definiert Anzahl Nachkommastellen

print('Gleitkommazahl: %5.2f' % (13.144))
```

```python
print('Gleitkommazahl: %5.0f' % (13.144))
```

```python
print('Gleitkommazahl: %1.5f' % (13.144)) # Auffüllen mit 0, wenn zuwenige Nachkommastellen
```

```python
print('Gleitkommazahl: %25.2f' % (13.144)) # Einrückung durch aufgefüllte Leerstellen
```

```python
# Kombinationen mehrerer Formatierungsoperatoren ist erlaubt.
print('Erster: %s, Zweiter: %5.2f, Dritter: %r' % ('Hallo!', 3.1415, 'Adios!'))
```

<h2 style='color: 	#b78900;'>Formatierung mit der Methode <b>.format()</b></h2>

```python
'Erst kommt {} etwas rein und dann {}'.format('HIER', 'DORT')
```

```python
# format() ist bessere Methode zum Einfügen in Print-Anweisungen
print('Dies ist ein String mit einer {}'.format('Einfügung'))
```

<h2 style='color: 	#b78900;'>Vorteile der Methode <b>.format()</b></h2>

```python
# Einfügung durch Indizierung
print('Max ist {2}, {0} und {1}'.format('schön', 'stark', 'groß'))
```

```python
# Variablen können eingefügt werden - ! WICHTIG : Definition in format()
print('Erstes Objekt: {a}, Zweites Objekt: {b}, Drittes Objekt: {c}'.format(
    a=1, b='Zwei', c=12.3))
```

```python
# Funktioniert nicht - bei Modulo war das anders
a, b, c = 1, 'Zwei', 12.3
p = 'Erstes Objekt: {a}, Zweites Objekt: {b}, Drittes Objekt: {c}'.format(a,b,c)

#*******************************
# Variablennamen übergeben
#a = "einen"
#b = "weiteren"
#print("Ich werde hier %s Text einfügen und einen %s hier." % (a, b))
```

```python
# Elemente können wiederholt werden
print('Ein gesparter %s ist ein verdienter %s.' % ('Cent', 'Cent'))
# vs.
print('Ein gesparter {c} ist ein verdienter {c}.'.format(c='Cent'))
```

<h2 style='color: 	#b78900;'>Formatierte Stringliterale (f-strings)</h2>

```python
# Direktes Einbinden von Variablen
name = 'Elke'
print(f"Heute hat {name} Küchendienst.")
```

```python
# !r liefert Stringrepräsentation
print(f"Heute hat {name!r} Küchendienst.")
```

<h2 style='color: 	#b78900;'>Gleitkommazahlformatierung mittels <br>"result: {value:{width}.{precision}}"</h2>

```python
# Gleitkommazahlformatierung
# Der Ausdruck `{value:10.4f}` der Methode `.format()` entspricht dem f-String `{value:{10}.{6}}`:
num = 78.45678
print(
    "Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {0:10.4f}".format(num))
print(
    f"Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {num:{10}.{6}}")
# Präzision bezieht sich auf gesamte Zahl
# Auffüllen von rechts, wenn Präzision das erlaubt, ist nicht möglich
```

```python
num = 12.456778
print("Zahl: {0:5.2f}".format(num))
```

```python
# Anwendung der format()-Syntax im f-String ist möglich
num = 35.77
print(
    "Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {0:10.4f}".format(num))
print(
    f"Dies ist meine vierstellige Gleitkommazahl mit zehn Zeichen: {num:10.4f}")

# Dann klappt es auch mit dem Auffüllen von rechts.
```
