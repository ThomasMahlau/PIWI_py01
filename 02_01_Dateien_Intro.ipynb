```python
"""
Der einfache Weg zum Öffnen einer Datei

fobj = open("woerterbuch.txt", "r")
# Befehle
fobj.close()

Besonderheiten:
- Datei öffnen
- Datei schließen
- Parameter                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                          spezifiziert, in welchem Modus auf die Datei zugegriffen werden kann
    r = read
    w = write
    ...
- FileNotFoundError, wenn Datei nicht an angegebener Stelle gefunden wird 
  (relative und absolute Pfade möglich)
- Datenströme stdin und stdout
    Eingabe erfolgt über Datenstrom stdin (input())
    Ausgabe erfolgt über Datenstrom stdout (print())


Inhalt der Datei woerterbuch.txt:
Spain Spanien
Germany Deutschland
Sweden Schweden
France Frankreich
Italy Italien
"""

fobj = open("woerterbuch.txt", "r")
# Befehle
fobj.close()

"""
BESSERE VARIANTE (automatisches Schließen)

with open("woerterbuch.txt", "r") as fobj:
    Befehle
"""

with open("woerterbuch.txt", "r") as fobj:
    for line in fobj:
        print(line)
```

```python
"""
Neuen Eintrag erfassen - ohne Überschreiben der Originaldatei
Variante I:
"""
woerter = {}
with open("woerterbuch.txt", "r") as fobj:
    for line in fobj:
        zuordnung = line.split(" ")
        if len(zuordnung) == 2:
            woerter[zuordnung[0]] = zuordnung[1].strip()
            
neu_engl = input("Neu Englisch")
neu_deutsch = input("Neu Deutsch")

with open("ausgabe.txt", "w") as fobj:
    for engl in woerter:
        fobj.write(f"{engl} {woerter[engl]}\n")
    fobj.write(f"{neu_engl} {neu_deutsch}")

with open("ausgabe.txt", "r") as fobj:
    for line in fobj:
        print(line)
```

```python
"""
Datei erweitern, statt überschreiben

BEACHTEN: Letzte Zeil muss leer sein, damit neuer Eintrag in neuer Zeile erscheint
"""

neu_engl = input("Neu Englisch")
neu_deutsch = input("Neu Deutsch")

with open("woerterbuch.txt", "a") as fobj:
    fobj.write(f"{neu_engl} {neu_deutsch}\n")

with open("woerterbuch.txt", "r") as fobj:
    for line in fobj:
        print(line)
```

# Parameter in der Übersicht

<table>
<tr><td>"r"</td><td>zum Lesen öffnen</td></tr>
<tr><td>"w"</td><td>zum Schreiben öffnen - vorhandene Datei wird überschrieben</td></tr>
<tr><td>"a"</td><td>zum Schreiben öffnen - vorhandene Datei wird  NICHT überschrieben</td></tr>
<tr><td>"x"</td><td>zum Schreiben öffnen - wenn bereits vorhanden FileExistsError-Exception</td></tr>
<tr><td>"r+", "w+", "a+", "x"</td><td>zum Lesen & Schreiben öffnen - wenn bereits vorhanden, wird Inhalt geleert</td></tr>
</table>

```python
"""
Daten aus einer Liste in eine Datei schreiben

Sie haben folgende Namensliste <namen>:
name = [
    "Peter Bauer",
    "Karl Schmidt",
    "Elvira Everglade"
]

Diese sollen in die mitarbeiternamen.csv mit einer Überschrift "Name,Vorname" und entsprechender 
Datenaufteilung eingetragen werden.
"""
```

```python
namen = [
    "Peter Bauer",
    "Karl Schmidt",
    "Elvira Everglade"
]

with open("mitarbeiternamen.csv", "w") as maliste:
    maliste.write("Name,Vorname\n")
    for name in namen:
        name = name.split(" ")
        vorname = name[0]
        nachname = name[1]
        maliste.write(f"{nachname},{vorname}\n")

with open("mitarbeiternamen.csv", "r") as maliste:
    for line in maliste:
        print(line.strip())
```
