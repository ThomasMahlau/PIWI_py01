```python
"""
Erstellen Sie eine Dictionary <begriffe> mit folgenden Inhalten:

Evaluation      Überprüfung eines Programm
Verifikation    Überprüfung, ob das Programm richtig funktioniert
Validierung     Überprüfung, ob das richtige Programm programmiert wurde

Lassen Sie den User einen Suchbegriff eingeben, damit im Dictionary danach gesucht werden kann.
Wenn der Suchbegriff gefunden wurde, soll die Erklärung ausgegeben werden, sonst soll eine entsprechende Meldung 
angezeigt werden.
"""
begriffe = {
    'Evaluation': 'Überprüfung eines Programms',
    'Verifikation': 'Überprüfung, ob das Programm richtig funktioniert',
    'Validierung': 'Überprüfung, ob das richtige Programm programmiert wurde'
}

def such(suche):
    """ Suche einen Begriff im Dictionary und gebe den gefundenen Wert zurück """
    if suche in begriffe:
        return begriffe[suche]
    else:
        return f"{suche} nicht gefunden"

suche = input("Suchbegriff eingeben:")
gefunden = such(suche)
print(gefunden)
```
