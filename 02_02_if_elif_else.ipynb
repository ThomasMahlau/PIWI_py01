<h1 style='color:#4876FF'><b>if, elif und else Anweisungen</b></h1>

Das if-Konstrukt erlaubt die Überprüfung von Bedingungen und das Auslösen von unterschiedlichen Aktionen in Abhängigkeit davon.

Allgemeine Syntax: 

    if Fall 1:
        erste Aktion
    elif Fall 2:
        zweite Aktion
    else:
        dritte Aktion

```python
# Einfache Überprüfung
if True:      # WICHTIG: Großschreibung !
    print('Es war wahr!')
```

```python
# Wenn - dann
x = False

if x:
    print('X war wahr!')
else:
    print('Ich werde in jedem Fall ausgegeben, in dem x nicht wahr ist')
```

```python
# Mehrere Überprüfungen

ort = 'Bank'

if ort == 'Autohaus':
    print('Willkommen im Autohaus!')
elif ort == 'Bank':
    print('Willkommen in der Bank!')
else:
    print('Wo bist du?')
```

<h1 style='color:#4876FF'><b>CASE in MATCH</b></h1>
<p>match startet ein <i>structural pattern matching</i> (ab 3.10)</p>
<p>Zu benutzen mit:</p>
<ul>
<li>Literalen (Zahlen, Zeichen)</li>
<li>OR oder |</li>
<li>Wildcard _</li>
<li>if-Konditionen für Sonderfälle</li>
</ul>


```python
status = 400
match status:
  case 400 | 404:
    # mit Sonderfallüberprüfung
    if status == 404:
      print("Fehlercode",404)
    else:
      print("Client Error")
  case 200:
    print("OK")
  case 500:
    print("Server Error")
```

```python
status = 201
match status:
  case 200:
    print("OK")
  case 201:
    print("Created")
  case _:
    print("Other")
```

```python
n = 10
match n:
  case 0 | 1:
    print("0 oder 1")
  case x if x % 2 == 0:
    print("Gerade Zahl:", x)
  case _:
    print("Ungerade Zahl:", n)
```
