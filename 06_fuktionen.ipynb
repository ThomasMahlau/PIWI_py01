# Funktionen

- Funktionen sollten immer oben platziert
- Schlüsselwort def funktionsname():
- best practise: nach Definitionsaufruf (def ...) ein Kommentar zur Aufgabe der Funktion
- Funktionen können Parameter aufnehmen: def funktionsname(param1, param2), die in der Funktion weiterverarbeitet werden
- Parameter können einzelne Werte sein oder Objekte (z.B.: Liste)
- Funktionen "kapseln" ihre Ergebnisse => diese stehen außerhalb der Funktion NICHT zur Verfügung
- Nur über ein 'return' wird das Ergebnis der Funktion an die aufrufende Ebene zurückgegeben
- Der Rückgabewert wird in einer Variablen in der aufrufenden Ebene 'aufgefangen'

```python
# 'Hallo Welt' mit einer Funktion ausgeben

def gruss():
    print("Hallo Welt")
    
gruss()
```

```python
# 'Hallo ...' mit einer Funktioin ausgeben

def gruss_person(name):
    print("Hallo", name)
    
gruss_person('Peter')
```

```python
# Funktion mit mehreren Parametern

def gruss_personen(name1, name2):
    print("Hallo", name1, 'und', name2)
    
gruss_personen('Peter', 'Mary')
```

```python
# Übergeben von Objekten als Parameter

def namen(namenliste, namen2):
    for name in namenliste:
        print(name)
    for name in namen2:
            print(name)
        
namensliste = ['Peter', 'Paul', 'Mary']
namen2 = ['Klaus', 'Sabine', 'Judith']

namen(namensliste, namen2)
```

```python
# Addieren von 2 Zahlen

def addiere(a, b):
    ergebnis = a + b
    print("Funktion", ergebnis)
    return ergebnis
    
ergebnis = addiere(2,2)

print("Aufrufende Ebene", ergebnis)
```
