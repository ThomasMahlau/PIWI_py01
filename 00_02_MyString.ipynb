```python
# Ziel: Erstes Wort ausgeben lassen
my_string = "Hallo Welt"

# 1. Variante
splitted = my_string.split(' ')
print(splitted[0])

# 2. Variante
gesuchter_index = my_string.find(' ')
print(my_string[0:gesuchter_index])

# 3. Variante
print(my_string[0:my_string.find(' ')])
```

# Ausgabeformatierung

```python
# Formatted-String
sprache = "Python"
bewertung = "wunderbar"
# bisher
message = sprache + " ist " + bewertung
print(message)

message2 = f"{sprache} ist {bewertung}"
print(message2)
```

```python
name = "Peter"
alter = 24
message = name + " ist " + str(alter)
print(message)
message2 = f"{name} ist {alter}"  # automatische Typumwandlung
print(message2)
```

```python
# Format()-Funktion
message ="Hallo {} {}".format("Peter!", "Du bist 24 Jahre alt")
print(message)
```

```python
ich = "Ich bin Alex. Ich bin "
eigenschaft1 = "gross"
eigenschaft2 = "klug"
eigenschaft3 = "schön"

message = "{} {} {} {}".format(ich, eigenschaft1, eigenschaft2, eigenschaft3)
print(message)
message2 = "{0} {3} {1} {2}".format(ich, eigenschaft1, eigenschaft2, eigenschaft3)
print(message2)
```
