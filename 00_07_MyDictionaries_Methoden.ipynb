### d.clear()

```python
d = {"k1": "v1", "k2": "v2", "k3": "v3"}
d
```

```python
d.clear()
d
```

### d.copy()

```python
d = {"k1": "v1", "k2": "v2", "k3": "v3"}
e = d.copy()
e
```

<p style='color:red;'>Beachten Sie, dass zwar das Dictionary selbst kopiert wird, es sich aber bei den Werten nach wie vor um <strong>Referenzen</strong> auf dieselben Objekte handelt.</p>
Siehe das folgende Beispiel:

```python
d1 = {"key": [1,2,3]}
d2 = d1.copy()
d2["key"].append(4)
d2
```

```python
"""Obwohl der Wert 4 in das kopierte Wörterbuch d2 hinzugefügt wurde,
enthält auch d1 den Wert 4, da beide Wörterbücher auf dasselbe Dictionary verweisen.
"""
d1
```

```python
# Der Vergleich mit dem is-Operator zeigt, dass beide Wörterbücher die gleichen Objekte haben.
d1['key'] is d2['key']
```

```python
# Löschen des Elements auf Index 3 der Liste, die dem Schlüssel "key" im Wörterbuch d2 zugeordnet ist.
del d2["key"][3]
```

```python
d1['key'] is d2['key']
```

### d.get(k, [x])
k = key<br>
x = Standardwert, der zurückgegben wird, wenn key nicht vorhanden ist. Es gibt keine Exception / Fehlermeldung.

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
gesuchert_key = "k2"
x = "Standardwert"

if gesuchert_key in my_dict:
    wert = my_dict[gesuchert_key]
else:
    wert = x
print(wert)
```

```python
# Mit der get-Methode
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
d.get("k2", 1337)
```

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
d.get("k5", 1337)
```

### d.items()
... gibt ein iterierbares Objekt über alle <strong>Schlüssel-Wert-Paare</strong> des Dictionarys in Form von <strong>Tupeln</strong> zurück.

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
for paar in d.items():
    print(paar)
```

### d.keys()
... gibt iterierbares Objekt (Liste) über alle Schlüssel des Dictionarys zurück.

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
for key in my_dict.keys():
    print(key)
print(my_dict.keys())
```

```python
# Was auch geht ...
for key in my_dict:
    print(key)
```

```python
# Schreiben der keys in eine Liste
mylist = list(my_dict.keys())
print(mylist)
```

### d.values()
... verhält sich wie <i>d.keys()</i>.

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
for value in my_dict.values():
    print(value)

print(my_dict.values())
```

<hr>

### Zwischenfazit keys - values - items
1. Wenn `keys & values` genutzt werden sollen: <strong>d.items()</strong>
2. Wenn nur `keys` genutzt werden sollen: <strong>d.keys()</strong>
3. Wenn nur `values` genutzt werden sollen: <strong>d.values()</strong>

### d.pop(k)
... löscht das Schlüssel-Wert-Paar mit dem Schlüssel k.

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
my_dict.pop("k1")
my_dict
```

### d.popitem()
... gibt ein willkürliches (aber nicht zufälliges) Schlüssel-Wert-Paar <strong>als Tupel</strong> und entfernt es aus dem Dictionary.<br>Ist das Dictionary leer, wird eine Exception erzeugt.<br>
Es könnte stets das zuletzt eingefügt Schlüssel-Wert-Paar sein.

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
my_dict.popitem()
```

### d.setdefault(k, [x])
k = key<br>
x = Standardwert, wenn kein Wert vorhanden

```python
my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
my_dict.setdefault("k2", 1337)
```

```python
# Da "k5" nicht existiert, wird der Schlüssel mit dem Wert 1337 hinzugefügt
my_dict.setdefault("k5", 1337)
my_dict
```

### d.update(d2)
... erweitert das Dictionary d um Schlüssel und Werte aus d2.<br>
Existiert der Schlüssel aus d2 bereits in d, werden die Werte von d überschrieben.

```python
# d.update() fügt hier neue Schlüssel-Wert-Paare hinzu

my_dict = {"k1": "v1", "k2": "v2", "k3": "v3"}
my_dict.update({"k4": "neuer Wert"})
my_dict
```

```python
# d.update() kann auch vorhandene Schlüssel-Wert-Paare überschreiben

my_dict.update({"k1": "überschriebener Wert"})
my_dict
```
