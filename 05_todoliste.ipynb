```python
def todos_listen():
    clear_output()
    if len(todoliste) == 0:
        print("Todoliste ist leer")
    else:
        for todo in todoliste:
            print(todo)

def todos_new():
    clear_output()
    newtodo = input("Neues Todo eingeben")
    todoliste.append(newtodo)

def todos_delete():
    clear_output()
    x = 0
    while x < len(todoliste):
        print(x, todoliste[x])
        x += 1
    loesche = int(input("Was möchtest du löschen? Ziffer eingeben!"))
    x = 0
    while x < len(todoliste):
        if x == loesche:
            todoliste.pop(x)
        x += 1
    x = 0
    clear_output()
    # Liste ohne gelöschtes Element
    while x < len(todoliste):
        print(x, todoliste[x])
        x += 1      
        
                  
menu = """
1 - Todos anzeigen
2 - Neues Todo erfassen
3 - Erledigt markieren
0 - Anwendung beenden
"""

from IPython.display import clear_output
clear_output()

todoliste = ['Bananen einkaufen', 'Blumen gießen']

while True:
    print(menu)
    auswahl = input("Bitte Menüpunkt wählen")
    if auswahl == '1':
        todos_listen()
    elif auswahl == '2':
        todos_new()
    elif auswahl == '3':
        todos_delete()
    elif auswahl == '0':
        clear_output()
        print("Programm wird beendet!")
        break
    else:
        clear_output()
        print("Falschen Menüpunkt eingegeben!!!")
```
