<h2 style='color: 	#b78900;'>Einfache For-Schleife</h2>

Schleifen können auf alle iterierbare Datenstrukturen angewendet werden:<br>
- Strings
- Listen
- Tupel
- Sets
- Dictionaries

```python
# Einen String durchlaufen
string = "Mein langer Text, der durchlaufen werden soll."
for s in string:
    print(s)
```

```python
# Eine Liste durchlaufen
names = ["Peter", "Paul", "Mary"]
for name in names:
    print(name)
```

<h2 style='color: 	#b78900;'>Wertgesteuerte For-Schleife</h2>

```python
names = ["Peter", "Paul", "Mary"]
for i in range(0, len(names)):
    # print(i)
    # print(i, names)
    print(i, names[i])
    
```

<h2 style='color: 	#b78900;'>While-Schleife mit Zähler</h2>

```python
names = ["Peter", "Paul", "Mary"]
x = 0
while x < len(names):
    print(names[x])
    x += 1
```

<h2 style='color: 	#b78900;'>While-Schleife mit True und Bedingung</h2>

```python
names = ["Peter", "Paul", "Mary"]

x = 0
anzahl = len(names)
while True:
    if x != anzahl:
        print(names[x])
        x += 1
    else:
        break
```
