### Strings

```python
my_string = "abc"
print(my_string)
new_string = my_string + "d"
print(new_string)
print(my_string)
my_string = my_string + "d"
print(my_string)
```

```python
# Indexing
erster_buchstabe = my_string[0]
print(erster_buchstabe)
```

```python
# Anzahl der Elemente des Strings ermitteln - Beispiel für eine Funktion / Methode für Strings
anzahl = len(my_string)
print(anzahl)
```

### Listen

```python
my_list = ["a", "b", "c", "d"]
print(my_list)
```

```python
# Hinzufügen eines weiteren Elements
my_list.append("e")
print(my_list)
```

```python
anzahl = len(my_list)
print(anzahl)
```

```python
my_string = "Peter, Paul, Mary"
print("String", len(my_string))

my_list = ["Peter", "Paul", "Mary"]
print("Liste", len(my_list))
```

```python
print(my_list[0])
```

### Tuple

```python
my_tuple = ("Peter", "Paul", "Mary")
print(my_tuple)
```

```python
print("Datentyp", type(my_tuple))
```

```python
print("Anzahl Elemente", len(my_tuple))
```

### Set

```python
my_set = {"Peter", "Paul", "Mary"}
print(my_set)
my_set = {"Peter", "Paul", "Mary", "Peter"}
print(my_set)
```

```python
anzahl = len(my_set)
print(anzahl)
```

```python
print("Datentyp", type(my_set))
```

### Dictionary

```python
my_dict = {'Name': 'Schuster', 'Vorname': 'Peter'}
print(my_dict)
```

```python
print(my_dict['Vorname'])
```

```python
mitarbeiter = {
     100 : { 'Name': 'Schuster', 'Vorname': 'Peter'},
     101 : { 'Name': 'Schneider', 'Vorname': 'Paul'},
     102 : { 'Name': 'Schmidt', 'Vorname': 'Mary'}
}

print(mitarbeiter[101])
print(mitarbeiter[101]['Name'])
```

### Mehrdimensionale Liste

```python
mitarbeiter = [
    [100, 'Schuster', 'Peter'],
    [200, 'Schneider', 'Paul'],
    [300, 'Schmidt', 'Mary']
]
```

```python
print(mitarbeiter[1][1])
```
