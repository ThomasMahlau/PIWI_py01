<h2 style='color:	#b78900;'>Mehrere Strings verbinden</h2>

```python
# z.B. aus einer Liste von Werten
students = ["Peter", "Paul", "Mary"]
print(students) # gibt Liste aus
print(', '.join(students)) # gibt Werte getrennt durch Komma aus

my_string = ', '.join(students)
print(my_string)
```

```python
# Zwei Listen zu einer machen

list1 = ["Peter", "Paul", "Mary"]
list2 = ["Karl", "Otto", "Claudia"]

liste = list1 + list2
print(liste)

# Kombinierte Liste in String umwandeln mit .join()
string_liste = ', '.join(liste)
print(string_liste)
```

```python
# Leerzeichen entfernen
string_liste.strip()
print(string_liste)
```

```python
# Anzeichen angegebenes Zeichen "C" in String ermitteln - ACHTUNG: case-sensitiv !!!
string_liste.count("C")
```

<h2 style='color:	#b78900;'>String-Slicing mit Hilfe des Slicing-Operators</h2>

```python
# Anzeigen der Zeichen vom Index 1 bis 5
mein_string = "Hallo Welt!"
mein_string[1:5]
```

```python
# Anzeigen des ersten Wortes (Index 0 bis 5)
mein_string[0:5]
```

```python
# gleiches Ergebnis (Index 0 bis 5)
mein_string[:5]
```

```python
# Alles aus dem String anzeigen
mein_string[:]
```

```python
# Letztes Zeichen ausgeben
mein_string[-1]
```

```python
# Gesamten String - außer letztes Zeichen
mein_string[:-1]
```

```python
# Gebe alles aus - in 1er-Schritten - von hinten (negatives Vorzeichen!)
mein_string[::-1]
```

```python
# Gebe alles aus - von vorne bis hinten - in 2er-Schritten 
mein_string[::2]
```

```python
# Gebe alles aus - in 3er-Schritten - von hinten
mein_string[::-3]
```

<h2 style='color: 	#b78900;'>Eingebaute grundlegende String-Methoden</h2>
<table>
<tr><td>.upper()</td><td>alles in Großbuchstaben</td></tr>
<tr><td>.lower()</td><td>alles in Kleinbuchstaben</td></tr>
<tr><td>.capitalize()</td><td>Anfangsbuchstabe des Strings als Großbuchstabe</td></tr>
<tr><td>.title()</td><td>Anfangsbuchstabe eines jeden Elements als Großbuchstabe</td></tr>
<tr><td>.split()</td><td>Umwandlung des Strings in eine Liste an Trennzeichen</td></tr>
<tr><td>.strip()</td><td>Abschneiden Leerzeichen am Anfang und Ende</td></tr>
<tr><td>.count()</td><td>Zählen gesuchtes Zeichen</td></tr>
<tr><td>.replace()</td><td>Ersetzen gesuchte(s) Zeichen(folge) durch neue(s)</td></tr>
<tr><td>.find()</td><td>Indexwert für gesuchte(s) Zeichen(folge) ermitteln</td></tr>
</table>

#### Object.Methode(Parameter)

```python
mein_string = "Hallo Welt! Wie geht es Dir?"
mein_string
```

```python
# Großbuchstaben erzeugen
mein_string.upper()
```

```python
# Nur Anfang des Strings wird großgeschrieben
mein_string.capitalize()
```

```python
# Jedes Wort erhält einen Großbuchstaben als Anfangsbuchstaben
mein_string.title()
```

```python
# Kleinbuchstaben erzeugen
mein_string.lower()
```

```python
# String-Bestandteile an Leerzeichen in Elemente umwandeln
mein_string.split()
```

```python
# String an einem bestimmten Zeichen aufteilen (Zeichen ist nicht enthalten)
mein_string.split('W')
```

```python
# Ermitteln der Anzahl von Zeichen in einem String
mein_string.count("l")
```

```python
# Zeichen ersetzen mit .replace(zu ersetzendes Zeichen, neues Zeichen)
mein_string = "Hallo Welt. Wie geht es Dir?"
mein_string = mein_string.replace("e", "_")
print(mein_string)
```

```python
# Zeichen suchen - 1. Indexwert erhalten
mein_string = "Hallo Welt. Wie geht es Dir?"
mein_string.find("Wie")
```

<h2 style='color: 	#b78900;'>Ausgabeformatierung</h2>

```python
# Formatted-String
sprache = "Python"
bewertung = "wunderbar"
mein_string = f"{sprache} ist {bewertung}!"
print(mein_string)
```

```python
# Format()-Funktion
mein_string = "Hallo Welt! {}".format('Wie geht es Dir?', 'Mir geht es gut!')
mein_string
```

```python
mein_string = "Hallo Welt! {} {}".format('Wie geht es Dir?', 'Mir geht es gut!')
mein_string
```

```python
mein_string = "Hallo Welt! {1} {0}".format('Wie geht es Dir?', 'Mir geht es gut!')
mein_string
```
