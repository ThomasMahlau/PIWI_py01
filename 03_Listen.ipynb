- listenname = [Wert1, Wert2 ... ]                 `! eckige Klammern`
- Liste können unterschiedliche Objekte enthalten
- erweiterbar mit `+ [Wert]`(wie Strings)
- veränderbar (`.append()`, `.pop()`, `.sort()`, `.reverse()`)
- indiziert
- Zusammenfügen mehrerer Listen ist möglich
<br><br>
<p><strong>Wichtige Methoden</strong></p>

<table>
<tr><td>sort()</td><td>Sortiert Elemente der Liste</td></tr>
<tr><td>type(liste)</td><td>Gibt Objekttyp der Liste zurück</td></tr>
<tr><td>append(elem)</td><td>Hinzufügen eines Elements</td></tr>
<tr><td>extend(liste)</td><td>Hinzufügen aller Elemente einer anderen Liste</td></tr>
<tr><td>index(elem)</td><td>Ermittelt Indexwert eines Elements</td></tr>
<tr><td>max(liste)</td><td>Ausgabe des Elements mit dem größten Wert</td></tr>
<tr><td>min(liste)</td><td>Ausgabe des Elements mit dem kleinstem Wert</td></tr>
<tr><td>len(liste)</td><td>Ausgabe des Anzahl der Elemente</td></tr>
<tr><td>list(seq)</td><td>Umwandeln eines Tupels in eine Liste</td></tr>

```python
my_list = [1,2,3,4,5,6,6,6,7]

# 1. Größten Wert herauszufinden
print(max(my_list))
# 2. Den kleinsten Wert herauszufinden
print(min(my_list))
# 3. my_list in my_list02 zu kopieren
my_list02 = my_list.copy()
print(my_list02)
# 4. Alle 6en bis auf eine aus my_list beseitigen
my_list.remove(6)
my_list.remove(6)
print(my_list)
# 5. Alle Elemente aus my_list02 löschen
my_list02.clear()
print(my_list02)
```
