# String-Slizing

```python
# Ausgabe aller Zeichen
my_string = "Peter, Paul, Mary"
my_string[:]
```

```python
# Ausgabe der ersten 5 Zeichen
my_string[0:5]      # End-Indexwert ist EXKLUSIV
```

```python
my_string[:5]       # Alle Zeichen vom Anfang bis Indexwert 5 (Exklusiv)
```

```python
my_string[:-1]      # Alle Zeichen von Anfang an, ohne letztes Zeichen
```

```python
my_string[::-1]     # Alle Zeichen beginnend vom letzten Zeichen mit 1er Schrittweite
```

```python
my_string[::2]      # Alle Zeichen von Anfang bis Ende mit 2er Schrittweite
```

# Ausgabeformatierungen

```python
# Arten von Stringformatierungen
"""
1. Konkatenieren (+)
2. Formatted-String (f"..{}..")
3. Format-Methode ("String {}".Format(...)
4. Modulo-Zeichen
"""
```

```python
# Konkatenieren
string01 = "Erster String. "
string02 = "Zweiter String. "
gesamtstring = string01 + string02
print(gesamtstring)
```

```python
# Konkatenieren
string = "Peter"
zahl = 24
print(string, "ist", zahl, "Jahre alt")
print(string + " ist " + str(zahl) + " Jahre alt")  # !!! Typumwandlung
```

```python
# f-String
string = "Peter"
zahl = 24
print(f"{string} ist {zahl} Jahre alt")
```

```python
# Format-Methode
string = "Peter"
zahl = 24
print("{} ist {} Jahre alt".format(string, zahl))
print("{0} ist {1} Jahre alt".format(string, zahl))
```

```python
# !!! Variablendefinition in format(...)
print('{string} ist {zahl} Jahre alt'.format(string='Peter', zahl=25))
```

```python
# Funktioniert NICHT
string = "Peter"
zahl = 24
print('{string} ist {zahl} Jahre alt'.format(string, zahl))
```

```python
# Modulo-Zeichen
print("%s ist %s Jahre alt" % (string, zahl))
```
