# Datentypen

```python
# PRIMITIVE DATENTYPEN
"""
- Zeichen (alphanumerische Zeichen)
- Zahlen (Ganzzahlen - integer)
- Float (Nachkommazahlen)
- Boolean (true - false)
"""
```

```python
# Addieren
ergebnis = 100 + 200
print(ergebnis)
```

```python
# Multiplizieren
ergebnis = 100 * 200
print(ergebnis)
type(ergebnis)
```

```python
# Dividieren
ergebnis = 200 / 100
print(ergebnis)
type(ergebnis)
```

```python
# Integerdivision
ergebnis = 250 // 100
print(ergebnis)
type(ergebnis)
```

```python
# Booleans - True oder False ist case-sensitive
print(type(False))
```

```python
# Modulo - Den Rest einer Division ermitteln
ergebnis = 7 % 4
print("Restwert:", ergebnis)
```

```python
# Pontenz
ergebnis = 2 ** 3
print(ergebnis)
```

```python
# Wurzel ziehen
ergebnis = 4 ** 0.5
print(ergebnis)
```
