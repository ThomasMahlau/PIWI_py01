# Schleifen / Loops

Schleifen können auf alle iterierbare Datenstrukturen angewendet werden:<br>
- Strings
- Listen
- Tupel
- Sets
- Dictionaries

```python
my_string = "Peter, Paul, Mary"

for name in my_string:
    print(name)
```

```python
freunde = ["Peter", "Paul", "Mary"]

for freund in freunde:
    print(freund)
```

# Wertgesteuerte For-Schleife

```python
freunde = ["Peter", "Paul", "Mary"]

for i in range(0, len(freunde)):
    print(i, freunde[i])
```

```python
print(freunde[2])
```

# While-Schleife mit Zähler

```python
freunde = ["Peter", "Paul", "Mary"]

x = 0

while x < len(freunde):
    print(x, freunde[x])
    x += 1
```

# While-Schleife mit True und Bedingung

```python
freunde = ["Peter", "Paul", "Mary"]

x = 0
anzahl = len(freunde)

while True:
    if x != anzahl:
        print(x, freunde[x])
        x += 1
    else:
        print("Letzter x-Wert", x)
        break
```
