# *args

```python
def multiplizieren(*args):
    """ Multipliziere die Summe einer beliebigen Anzahl von Argumenten
    mit 0.05 und gib das Ergebnis zurück """
    
    ergebnis = sum(args) * 0.05
    return ergebnis

    # return sum(args) * 0.05
    
multiplizieren(10,10,10,100,1000,1)
```

```python
def multi(a=0,b=0,c=0,d=0,e=0,f=0):
    ergebnis = (a+b+c+d+e+f)*0.05
    return ergebnis

multi(10,10,10,100,1000)
```

```python
def multi_liste(my_list):
    ergebnis = sum(my_list) * 0.05
    return ergebnis
    
my_list = [10,10,10,100,1000,1]

multi_liste(my_list)
```

# **kwargs

ermöglicht die Übergabe von Schlüssel-Wert-Paaren in eine Funktion.

Beliebige Parameternamen sind möglich - aber: kwargs ist Konvention.

```python
def lieblingsfrucht(**kwargs):
    """ Überprüft, ob die Schlüsselwort-Eingabeparamenter den Schlüssel 'frucht' 
    enthält und gibt einen dazu passenden String aus.
    """
    
    if 'frucht' in kwargs:
        print(f"Meine Liebingsfrust ist {kwargs['frucht']}.")
    else:
        print("Ich mag keine Früchte!")
        
lieblingsfrucht(frucht='Ananas')
```

# Kombination aus *args und **kwargs

!!! *args müssen immer VOR den **kwargs stehen

```python
def meine_funktion(*args, **kwargs):
    summe = sum(args)
    if 'monat' in kwargs:
        print(f"Die Provision für den Monat {kwargs['monat']} beträgt {summe} Euro \
            für den Mitarbeiter {kwargs['mitarbeiter']}.")
    else:
        print("Es gibt keine Provision. Sorry.")
        
        
meine_funktion(100,150,300,monat = 'September',mitarbeiter='Hubert')
```
