<h2 style='color:	#b78900;'>String-Methoden</h2>
<table>
<tr><td>.lower()</td><td>Alle Zeichen werden klein geschrieben</td></tr>
<tr><td>.upper()</td><td>Alle Zeichen werden groß geschrieben</td></tr>
<tr><td>.strip()</td><td>Liefert Zeichenfolge ohne Leerzeichen am Anfang und Ende</td></tr>
<tr><td>.find('anderes_zeichen')</td><td>Sucht nach angegebenen Zeichen(folge) und gibt 1. Indexwert zurück oder -1, wenn nicht gefunden</td></tr>
<tr><td>.replace('old', 'new')</td><td>Ersetzt vorhandene Zeichen(folge) durch neue</td></tr>
<tr><td>.split('delim')</td><td>gibt eine Liste von Teilzeichenfolgen zurück, die durch angegebenes Trennzeichen separiert sind - wenn keine Angabe, dann wird Leerzeichen als Trenner angenommen</td></tr>
<tr><td>.join(list)</td><td>Verbindet angegebene Liste zu einem String</td></tr>
</table>

```python
my_string = "Peter, Paul, Mary"
print('Kleinbuchstaben:', my_string.lower())
print('Großbuchstaben:', my_string.upper())
```

```python
aktenbezeichnung = ["abc100", "fBG200", "aaa300"]
print(aktenbezeichnung[0].upper())
```

```python
aktenbezeichnung = "abc100, fBG200, aaa300"
wo_ist = aktenbezeichnung.find('aaa300')
print(wo_ist)
```

```python
print(aktenbezeichnung.replace('aaa300', 'AAA300'))
print(aktenbezeichnung)
```

### Umwandlung eines Strings in eine Liste mit .split()

```python
aktenbezeichnung = "abc100, fBG200, aaa300"
print("Elemente:", len(aktenbezeichnung))

my_list = aktenbezeichnung.split(', ')
print(my_list)
print("Elemente:", len(my_list))
```

```python
string_mit_leerzeichen = " Peter, Paul, Mary "
print(string_mit_leerzeichen.strip())
print(string_mit_leerzeichen)
```

```python
ersetze_leerzeichen = string_mit_leerzeichen.replace(' ', '_')
print(ersetze_leerzeichen)
```

```python
# Aus einem Namen eine Emailadresse generieren
# "Schuster, Paul"
# gewünschte Emailadresse: p.schuster@firma.de

my_string = "Schuster, Paul"
my_list = my_string.split(', ')
print(my_list)

vname =my_list[1][0]
print(vname)

email = (vname+"."+my_list[1]+'@firma.de').lower()
print(email)
```
