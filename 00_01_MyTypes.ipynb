<h1 style='color:#4876FF'><b>Datentypen / Datenstrukturen - Übersicht</b></h1>
<ul>
<li>Variablen mit "primitiven" Datentypen
    <ul>
        <li>Zeichen</li>
        <li>Zahlen (Int, Fload, Boolean)</li>
    </ul>
</li>
<li>"Dynamic Typing" (Einer Variable mit Wert in einem bestimmten Datentyp kann ohne weiteres ein anderer Wert eines anderen Datentyps zugewiesen werden)</li>
</ul>
<table>
    <tr><td>String</td><td style="width:300px;">my_string = "x, y, z"</td><td>Durch Hochkomma oder Anführungszeichen begrenzte Zeichen(folge), die Elemente durch Komma separiert</td><td>unveränderbar, erweiterbar, indizierbar</td><td>Methoden verfügbar</td></tr>
    <tr><td>Listen</td><td>my_list = ["x", "y", "z"]</td><td>Durch eckige Klammern begrenzte Liste von Elementen oder Objekten</td>
    <td>veränderbar, erweiterbar, indizierbar</td><td>Methoden verfügbar</td>
    </tr>
    <tr><td>Tuple</td><td>my_tuple = ("x", "y", "z")<td>sind Listen sehr ähnlich</td><td>unveränderbar, indizierbar</td><td>Methoden verfügbar</td></tr>
    <tr><td>Set</td><td>my_set = {"x", "y", "z"}</td><td>Ungeordnete (!) Sammlung einzigartiger (!) Elemente mit wechselnder Reihenfolge (!)</td><td>veränderbar, NICHT erweiterbar, NICHT indizierbar</td><td>Methoden verfügbar</td></tr>
    <tr><td>Dictionaries</td><td>mein_dict = {'key1': "Eins", 'key2': 2}</td><td>Sammlung von Objekten, die über einen Key(Schlüssel) gespeichert werden (Key-Value-Paare)</td><td>
    veränderbar, erweiterbar, ungeordnet, NICHT indizierbar</td><td>Methoden verfügbar</td></tr>
</table>

```python
# Typumwandlung String in Int
a = "1"
b = int(a)
print("a:", a)
print("b:", b)
print("Typ von b:", type(b))
```

```python
# Typumwandlung Int in String
a = 1
b = str(a)
print("a:", a)
print("b:", b)
print("Typ von b:", type(b))
```

```python
# Typumwandlung Float in Int
a = 1.7
b = int(a)
print("a:", a)
print("b:", b)
print("Typ von b:", type(b))
```

```python
# Typumwandlung Float in Int
a = 1
b = float(a)
print("a:", a)
print("b:", b)
print("Typ von b:", type(b))
```

```python
# Implizite Typumwandlung durch Division
a = 4
b = 2
c = a / b
print("a:", a)
print("b:", b)
print("c:", c)
print("Typ von c:", type(c))
```
