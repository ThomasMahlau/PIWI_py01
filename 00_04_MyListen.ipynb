```python
liste = [1, 2, 3, 4]
# Anzahl eines benannten Elements ermitteln
print("Häufigkeit 1: ", liste.count(1))

# Größten Wert suchen
max_value = max(liste)
print("Größter Wert: ", max_value)

# Kleinsten Wert suchen
min_value = min(liste)
print("Kleinster Wert: ", min_value)

# Kopieren einer Liste
liste = [1, 2, 3, 4]
liste2 = []
liste2 = liste.copy()
print("Kopierte Liste: ",liste2)

# Liste leeren
liste.clear()
print("Geleerte Liste:", liste)
```

```python
# Elemente ergänzen mit "+" oder "+=" oder .append()
import time
n= 100000
start_time = time.time()
l = []
for i in range(n):
    l = l + [i * 2]
print(time.time() - start_time)
```

```python
start_time = time.time()
l = []
for i in range(n):
    l += [i * 2]
print(time.time() - start_time)
#print(start_time,l)
```

```python
# Die schnellste Variante
start_time = time.time()
l = []
for i in range(n):
    l.append(i * 2)
print(time.time() - start_time)
```

```python
# Tupel anlegen
my_tupel = (100, 200, 300)
print(my_tupel)

# Umwandeln des Tupel in eine Liste
my_list = list(my_tupel)
print(my_list)
```

<h2 style='color: 	#b78900;'>Noch einmal: Grundlegende Methoden</h2>


### Die wichtigsten Methoden für Listen in der Übersicht

* append (anhängen)
* count (zählen)
* extend (erweitern)
* insert (einfügen)
* pop (extrahieren)
* remove (entfernen)
* reverse (invertieren)
* sort (sortieren)

```python
# Neue Liste erstellen
liste = [1, 2, 3]
```

#### Methode `append()`

```python
liste.append('Neues Item') # permanentes Hinzufügen
liste
```

### Methode `pop()`

```python
# Letztes Element löschen
liste.pop()
liste
```

```python
# Gezieltes Löschen eines Elements
liste.pop(0)
liste
```

```python
# Gelöschtes Element einer Variablen zuweisen - z.B. für User-Feedback
popped = liste.pop()
print("Gelöscht wurde: ",popped)
```

```python
liste
```

```python
# Fehler, wenn Index nicht vorhanden
geloescht = liste.pop(100)
```

### `del`

```python
# Listeneintrag löschen mit del - KEIN Rückgabewert möglich
students = ["Max", "Monika", "Erik", "Franziska", "ABCDEF"]
del students[3]
print(students)
```

### Methode `remove()`

```python
students = ["Max", "Monika", "Erik", "Franziska", "ABCDEF"]
students.remove("Monika")
print(students)
```

<h2 style='color: 	#b78900;'>Sortieren & invertieren</h2>

```python
neue_liste = ['a', 'x', 'o', 'b']
neue_liste
```

### Methode `reverse()`

```python
# Listenelemente umgekehrt anordnen - entspricht NICHT absteigender Sortierung
neue_liste.reverse()
neue_liste
```

### Methode `sort()`

```python
# Listelemente aufsteigend sortieren (Zeichen nach Alphabet, Zahlen aufsteigend)
neue_liste = ['a', 'x', 'o', 'b']
neue_liste.sort()
print(neue_liste)
```

```python
# Absteigende Reihenfolge
neue_liste = ['a', 'x', 'o', 'b']
neue_liste.sort(reverse=True)
print(neue_liste)
```

<h2 style='color: 	#b78900;'>Verschachtelte Listen</h2>

```python
# Variante 1 - Liste mit Einzelelementen aus 3 Listen
lst_1 = [1,2,3]
lst_2 = [4,5,6]
lst_3 = [7,8,9]

lst_1.extend(lst_2)
lst_1.extend(lst_3)
print(lst_1)

# Variante 2: Drei Listen erstellen - Verschachtelte Listen (drei Listen in einer Liste)
lst_1 = [1,2,3]
lst_2 = [4,5,6]
lst_3 = [7,8,9]

# Drei Listen zu einer machen
matrix = [lst_1, lst_2, lst_3]
print(matrix)
```

```python
# Erstes Element der Matrix
matrix[0]
```

```python
# Erstes Element des ersten Elements auslesen
matrix[0][0]
```
