<h1 style='color:#4876FF'><b>Vergleichsoperatoren</b></h1>

Vergleichsoperatoren erlauben es Variablen zu vergleichen und die Werte `True` oder `False` auszugeben.<br>
Bitte beachten: `Großschreibung` ist wichtig, wenn True oder False in der Codierung benutzt werden!<br>
Stellen wir uns für die nachfolgenden Beispiele vor:<br>
a = 3 und b = 5

<tr>

## <span style='color:	#b78900;'>Verknüpfte Vergleichsoperatoren</span>

Vergleichsoperatoren können miteinander verknüpft werden, um komplexere Überprüfungen vorzunehmen.
Die Verknüpfung kann außerdem mit Hilfe von `and` oder `or` erfolgen.

```python
1 < 2 < 3
```

```python
1 < 2 and 2 < 3
```

Das `and`-Statement stellt sich, dass zwei Bedingungen gleichzeitig `True` (wahr) sind.

```python
# gleiches Ergebnis - nur kürzer
1 < 2 < 3
```

Verwendung des `or`-Statements.<br>
Nur eine Bedingung muss wahr sein, damit True zurückgegeben wird.

```python
1 == 2 or 2 < 3
```
