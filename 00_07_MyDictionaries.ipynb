<h1 style='color:#4876FF'><b>Dictionaries</b></h1>

Dictionaries in Python entsprechen <b>Maps oder Hash-Tables</b>, die aus anderen Programmiersprachen bekannt sind.<br>
Eine `Map` in Python ist eine `Sammlung von Objekten`, die über einen `Key` (Schlüssel) gespeichert werden.<br><br>
`Merkmale`:
- Key-Value-Koppellung
- ungeordnet
- veränderbar
- Aufrufen der Werte über die Keys

```python
my_dict = {
    'wert1': 'Erklärung 1',
    'wert2': 'Erklärung 2',
}

gesucht = 'wert1'
if gesucht in my_dict:
    print('gefunden:',my_dict[gesucht])
else:
    print(my_dict[gesucht],'nicht gefunden')
```

```python
my_liste = ["Deutschland", "Frankreich", "Italien"]
gesucht = "Deutschland"
for x in my_liste:
    if x == gesucht:
        print(x)
```

```python
"""Strukturierte Speicherung von Daten"""
gesucht = "Germany"

woerterbuch = {
    'Germany': 'Deutschland',
    'France': 'Frankreich',
    'Italy': 'Italien'
}
for key, value in woerterbuch.items():
    if key == gesucht:
        print("Gefunden:",value)
```

```python
"""Items gezielt auslesen über Schlüssel & Index
"""
mein_dict = {
    'key1': 1223, 
    'key2': [1,2,3,4], 
    'key3': ['Item1', 'Item2', 'Item3']}
mein_dict['key3'][2]
```

```python
"""Methoden auf Item anwenden - nicht permanent
"""
mein_dict['key3'][2].upper()
mein_dict
```

```python
"""Werte der Items beeinflussen - z.B.: Rechnen mit Items (nicht permanent)
"""
mein_dict['key1'] - 10
mein_new_value = mein_dict['key1'] - 10
print(mein_new_value)
print(mein_dict)
```

```python
"""Permanente Veränderung durch Neuzuweisung und Speichern in Variable
"""
mein_dict = {'key1': 1223, 'key2': [1,2,3,4], 'key3': ['Item1', 'Item2', 'Item3']}
meine_berechnung = mein_dict['key1'] - 10
print("Variable",meine_berechnung)

# ODER durch Neuzuweisung des Wertes
mein_dict['key1'] = mein_dict['key1'] - 10
print("Neuzuweisung",mein_dict['key1'])
```

```python
"""Selbstaddition etc. ebenfalls möglich
"""
mein_dict = {'key1': 1223, 'key2': [1,2,3,4], 'key3': ['Item1', 'Item2', 'Item3']}
meine_berechnung = mein_dict['key1'] - 10
meine_berechnung += meine_berechnung
print(meine_berechnung)
```

```python
my_mitarbeiter = {
    '100': {'Name': 'Schulte', 'Vorname': 'Anna', 'Alter': 28, 'Abteilung': 'Vertrieb'},
    '101': {'Name': 'Müller', 'Vorname': 'Bernd', 'Alter': 35, 'Abteilung': 'IT'},
    '102': {'Name': 'Schneider', 'Vorname': 'Clara', 'Alter': 30, 'Abteilung': 'Marketing'}
}

"""Alle Mitarbeiter-Namen ausgeben
"""
for x in my_mitarbeiter.values():
    print(x['Name'])
```

```python
"""Weitere Mitarbeiter-Daten ausgeben lassen
"""
for x, y in my_mitarbeiter.items():
    print(x, y['Name'], y['Vorname'], y['Alter'], y['Abteilung'])
```

```python
my_array = {
    'Name': 'Anna',
    'Alter': 28,
    'Funktion': 'Entwicklerin',
    'Abteilung': 'IT'
}

""" Suche mit Key """
my_array['Funktion']
```

```python
""" Suche mit Value"""
my_multi_array = {
    '1': {'Name': 'Anna', 'Alter': 28, 'Funktion': 'Entwicklerin', 'Abteilung': 'IT'},
    '2': {'Name': 'Bernd', 'Alter': 35, 'Funktion': 'Manager', 'Abteilung': 'IT'},
    '3': {'Name': 'Clara', 'Alter': 30, 'Funktion': 'Marketing-Spezialistin', 'Abteilung': 'Marketing'}
}
print("Suche mit Values:",my_multi_array['2'].values())
print("Suche mit Key:",my_multi_array['2'])
```
