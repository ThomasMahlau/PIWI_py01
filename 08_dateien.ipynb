```python
# Öffnen einer Datei und Auslesen des Inhalts

with open("woerterbuch.txt", "r") as fobj:
    for line in fobj:
        print(line.strip())   # .strip() entfernt Sonderzeichen - i.d.F.: Zeilenumbrüche aus der Datei
```

```python
woerter = {}
with open("woerterbuch.txt", "r") as fobj:
    for line in fobj:
        zuordnung = line.split(", ")
        if len(zuordnung) == 2:
            woerter[zuordnung[0]] = zuordnung[1].strip()
            
#print(woerter)

#woerter['Germany']

while True:
    gesucht = input("Bitte Suchbegriff eingeben - q für Quit")
    if gesucht in woerter:
        print("Das deutsche Wort lautet:", woerter[gesucht])
    elif gesucht == 'q':
        break
    else:
        print("Das Wort ist unbekannt.")
        
```

```python
# Neuer Eintrag OHNE zu überschreiben

while True:
    englisch_new = input("Neu Englisch oder 'q' für Quit ")
    if englisch_new == 'q':
        break
    german_new = input("Neu Deutsch")

    with open("woerterbuch.txt", "a") as fobj:
        fobj.write(f"{englisch_new}, {german_new}\n")
        
    with open("woerterbuch.txt", "r") as fobj:
        for line in fobj:
            print(line.strip())
            
    
```
